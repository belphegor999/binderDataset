```python
import pandas as pd
import matplotlib.pyplot as plt
```

```python
df = pd.read_csv("temperaturas.csv")
df.head()
```

```python
plt.plot(df["dia"], df["temperatura"], marker="o")
plt.title("Temperatura diaria")
plt.xlabel("Día")
plt.ylabel("Temperatura (°C)")
plt.grid(True)
plt.show()
```
